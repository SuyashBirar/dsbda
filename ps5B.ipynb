```python
import matplotlib.pyplot as plt
mean_score = df.groupby('species')['PetalWidthCm'].mean()
plt.figure()
mean_score.plot(kind="bar")
plt.title("Average Mean PetalWidth by species of iris")
plt.xlabel("Species")
plt.ylabel("Average Petal Width in cm")
plt.plot()
```
