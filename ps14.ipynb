```python
import pandas as pd

df = pd.read_csv("C:/Users/Admin/Downloads/DSBDA Pr/sample_weathertxt.txt", sep=",")
print(df.head())
```

```python
#Convert Date Column
df["Date"] = pd.to_datetime(df["Date"])

# Extract Year & Month
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
```

```python
print(df.columns)
```

```python
#Year-wise Average
year_avg = df.groupby("Year")[["Temp", "DewPoint", "WindSpeed"]].mean()
print("Year-wise Average:\n", year_avg)
```

```python
#Month-wise Average
month_avg = df.groupby("Month")[["Temp", "DewPoint", "WindSpeed"]].mean()
print("Month-wise Average:\n", month_avg)
```

```python
#Year-Month Combined
year_month_avg = df.groupby(["Year", "Month"])[
    ["Temp", "DewPoint", "WindSpeed"]
].mean()

print("Year-Month Average:\n", year_month_avg)
```
