```python
import pandas as pd
import numpy as np

df = pd.read_csv("C:/Users/Admin/Downloads/studentsPerformance.csv")
print(df.head())
```

```python
df['mathscore'] = df['mathscore'].fillna(df['mathscore'].mean())
```

```python
df.describe()
```

```python
# groupby() method  group data by values
mean_score = df.groupby('gender')['mathscore'].mean()
mean_score
```

```python
df.groupby('gender')['mathscore'].median()
```

```python
df.groupby('gender')['mathscore'].min()
```

```python
df.groupby('gender')['mathscore'].max()
```

```python
df.groupby('gender')['mathscore'].std()
```

```python
# importing matplotib to plot a Bar Chart to show the mean of Male and Female
import matplotlib.pyplot as plt
plt.figure()
mean_score.plot(kind='bar')
plt.title("Average Math Score by gender")
plt.xlabel("Gender")
plt.ylabel("Mean Math Score")
plt.plot()
```

```python
#Convert Categorical Variable to Numeric List
# Using mapping
gender_map = {"female": 0, "male": 1}
df["gender_numeric"] = df["gender"].map(gender_map)

# Convert to list
gender_list = df["gender_numeric"].tolist()

print(gender_list[:10])  # first 10 values
```
